```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Algoritimo de calssficação --> Regressão logistica --> entrega a probabilidade das classes
# 0: STABLE -- 1: unstable
# registro A --> [0.98, 0.02] --> 0
# registro B --> [0.64, 0.36] --> 0
# registro c --> [0.49, 0.51] --> 1
from sklearn.linear_model import LogisticRegression
# Separação dos dados para treinamento e teste (train-test-split)
from sklearn.model_selection import train_test_split
#metrica de avaliação
from sklearn.metrics import accuracy_score
```

```python
dados = pd.read_csv('https://raw.githubusercontent.com/IagoNeiva/CP2-1CCPX/refs/heads/main/Data_for_UCI_named.csv')
dados.head()
```

```python
# shape, infos
```

```python
# Features
X = dados.drop(columns=['stab', 'stabf'], axis=1)
X.head()
```

```python
# Target (stable ou unstable)
y = dados['stabf']
y.head(10)
```

```python
# Separação de dados de treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=41)
```

```python
# Quantas classes temos na coluna alvo?
dados['stabf'].unique()
```

```python
# Quantos registros temos na coluna alvo?
dados['stabf'].value_counts()
```

```python
X.shape
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
modelo = LogisticRegression()
```

teste de modelo

```python
modelo.fit(X_train, y_train)
```

Gerando um array de probabilidades

```python
probabilidades = modelo.predict_proba(X_test)
probabilidades[:50]
```

Gerando array de previsões

```python
y_predict = modelo.predict(X_test)
y_predict[:50]
```

```python
# Acurácia
acuracia = 100 * accuracy_score(y_test, y_predict)
print(f'A acurácia do modelo baseado em Regressão Logística é: {acuracia:.2f}%')
```
